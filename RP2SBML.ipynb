```python
# Fixing E. coli vio file

import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import numpy as np
from sklearn.metrics import r2_score

import data
import utils
import plot
import warnings
import cobra

warnings.filterwarnings("ignore", category=UserWarning, module="cobra")

folder    = './'
cobra_model_file = folder + 'data/' + 'iML1515_VIO_duplicated.xml'
cobra_model_save = folder + 'data/' + 'iML1515_VIO_duplicated_fixed.xml'
biomass_rxn_id   = 'BIOMASS_Ec_iML1515_core_75p37M'

model = cobra.io.read_sbml_model(cobra_model_file)   # fresh model

# -----------------------------
# Rename reaction rxn_target -> EX_rxn_target
# -----------------------------
rxn_target = model.reactions.get_by_id("rxn_target")
rxn_target.id = "EX_TARGET_o"
rxn_target.name = "EX_TARGET_o"

# -----------------------------
# Save the modified model
# -----------------------------
cobra.io.write_sbml_model(model, cobra_model_save)

# -----------------------------
# Reload and check
# -----------------------------
test_model = cobra.io.read_sbml_model(cobra_model_save)

met_check = test_model.metabolites.get_by_id("TARGET")
rxn_check = test_model.reactions.get_by_id("EX_TARGET_o")

print("Saved file:", cobra_model_save)
print("Metabolite id  :", met_check.id)
print("Metabolite name:", met_check.name)
print("Reaction id    :", rxn_check.id)
print("Reaction name  :", rxn_check.name)
print("Metabolites    :", len(test_model.metabolites))
print("Reactions      :", len(test_model.reactions))
```
